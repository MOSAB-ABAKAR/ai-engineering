```python
#%%  import libraries and load the environment variables
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
if OPENAI_API_KEY is None:
    raise Exception("Your Openi AI key is missing. please add it to your .env file")
```

```python
#%%  add  System message to the chat
system_message = " Your are a web developer and you have to answer the user questions in a simple way and provide code examples when needed.    "


user_prompt = """
Help me to create a simple HTML page with a header, a paragraph, and a button that changes the text of the paragraph when clicked.  

1. My header should say "Welcome to My Page"
 - The paragraph should say "This is a simple HTML page."
 - The button should say "Click Me!" and when clicked, it should change the paragraph text to "You clicked the button!"
 - the page should have a light blue background color and the text should be centered on the page.


2. Please provide the complete HTML code for this page, including the necessary CSS and JavaScript to achieve the desired functionality.
 - paragraph text should be centered and have a font size of 18px.
 - two buttons should be added, one to change the paragraph text and another to reset it back to the original text.
 -  clear comments should be added to the code to explain each section and functionality.

3. home page should have a navigation bar with links to "Home", "About", and "Contact" sections.
  - each section should have a unique ID and a brief description of what the section is about.
  - footer containing copyright information and social media links should be added at the bottom of the page.
  - logo image should be included in the header section, and it should be responsive to different screen sizes.


"""

requiremnts_list = " qualifications, skills, and experience required for a web developer position.  "


company_policy = " company policies, work culture, and benefits offered to employees.  "

#import client OpenAI
client = OpenAI(api_key=OPENAI_API_KEY)
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_prompt},
    ],
)

display(Markdown(completion.choices[0].message.content))
```

```python
career_suggestions = "completion.choices[0].message.content"

reasoning_prompt = """

help me to provide a detailed reasoning for the career suggestions provided by the AI model. {career_suggestions}


Answer according to the user prompt and provide a detailed reasoning for the career suggestions provided by the AI model.
{user_prompt}

Do not transgress the requirements and the company policy provided by the user. "{requiremnts_list}"
 this is the company policy: "{company_policy}"

"""

#%%  create a client and send the request to the OpenAI API

completion = client.chat.completions.create(
    model="gpt-o3-mini",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": reasoning_prompt},
    ]
    max_tokens=1000,
    temperature=0.7,
)

#%%  display the response from the OpenAI API
response = completion.choices[0].message.content
display(Markdown(response))
```
